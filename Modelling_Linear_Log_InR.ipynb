Build a Linear Regression model and a Logistic Regression model using the given
dataset. Evaluate both models using the test data. Use the Linear Regression model to
predict numerical quantities and use the Logistic Regression model to predict the
class/category of new observations.

```r
library(readr)

gym_data <- read_csv("/content/gym_regression_classification_dataset_160.csv")
```

```r
gym_data <- read.csv("/content/gym_regression_classification_dataset_160.csv")

head(gym_data)
str(gym_data)
summary(gym_data)
```

Preprocessing

```r
gym_data$Exercise_Type <- as.factor(gym_data$Exercise_Type)
gym_data$Fitness_Level <- as.factor(gym_data$Fitness_Level)

colSums(is.na(gym_data))
```

Sampling for Modeling and Validation

```r
set.seed(123)

train_index <- sample(
  1:nrow(gym_data),
  0.8 * nrow(gym_data)
)

train_data <- gym_data[train_index, ]
test_data <- gym_data[-train_index, ]

nrow(train_data)
nrow(test_data) #Split into training and testing data
```

**Linear Regression**

```r
linear_model <- lm(
  Weekly_Expenditure ~ Height_cm + Weight_kg + BMI +
    Calorie_Intake + Exercise_Type,
  data = train_data
)

summary(linear_model)
```

Predict numerical quantities

```r
linear_pred <- predict(
  linear_model,
  test_data
)

head(linear_pred)
```

```r
rmse <- sqrt(
  mean(
    (test_data$Weekly_Expenditure - linear_pred)^2
  )
)

rmse #Evaluate the Linear Regression model
```

```r
r2 <- cor(
  test_data$Weekly_Expenditure,
  linear_pred
)^2

r2
```

```r
data.frame(
  Actual = test_data$Weekly_Expenditure,
  Predicted = round(linear_pred, 2)
)
```

**-----------Logistic Regression-----------**

```r
logistic_model <- glm(
  Fitness_Level ~ Height_cm + Weight_kg + BMI +
    Calorie_Intake + Exercise_Type,
  data = train_data,
  family = binomial
)

summary(logistic_model)
```

```r
prob <- predict(
  logistic_model,
  test_data,
  type = "response"
)

predicted_class <- ifelse(
  prob >= 0.5,
  "Fit",
  "Needs_Improvement"
)

predicted_class <- factor(
  predicted_class,
  levels = levels(test_data$Fitness_Level)
)

head(predicted_class)
```

```r
confusion_matrix <- table(
  Actual = test_data$Fitness_Level,
  Predicted = predicted_class
)

confusion_matrix
```

```r
accuracy <- mean(
  predicted_class == test_data$Fitness_Level
)

accuracy
```

```r
TP <- confusion_matrix["Fit", "Fit"]
FP <- confusion_matrix["Needs_Improvement", "Fit"]

precision <- TP / (TP + FP)

precision
```

```r
FN <- confusion_matrix["Fit", "Needs_Improvement"]

recall <- TP / (TP + FN)

recall
```

Predict a New Observation

```r
new_member <- data.frame(
  Height_cm = 175,
  Weight_kg = 72,
  BMI = 23.5,
  Calorie_Intake = 2600,
  Exercise_Type = factor(
    "Hypertrophy",
    levels = levels(gym_data$Exercise_Type)
  )
)
```

```r
new_expenditure <- predict(
  linear_model,
  new_member
)

new_expenditure
```

```r
new_probability <- predict(
  logistic_model,
  new_member,
  type = "response"
)

new_class <- ifelse(
  new_probability >= 0.5,
  "Fit",
  "Needs_Improvement"
)

new_probability
new_class
```
