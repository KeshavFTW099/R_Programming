**LINEAR REGRESSION**

```r

car_data <- read.csv("/content/car_regression_classification_dataset_160.csv")

head(car_data)
str(car_data)
summary(car_data)

#Preprocessing
car_data$Brand <- as.factor(car_data$Brand)
car_data$Fuel_Type <- as.factor(car_data$Fuel_Type)

#missing values
colSums(is.na(car_data))

#data into training and testing sets
set.seed(123)

train_index <- sample(
  1:nrow(car_data),
  size = 0.8 * nrow(car_data)
)

train_data <- car_data[train_index, ]
test_data <- car_data[-train_index, ]

#Linear Regression model
linear_model <- lm(
  Price_Lakh ~ Engine_CC + Mileage_kmpl +
    Car_Age + Brand + Fuel_Type,
  data = train_data
)

#Model
summary(linear_model)


predicted_price <- predict(
  linear_model,
  newdata = test_data
)

#RMSE
rmse <- sqrt(
  mean(
    (test_data$Price_Lakh - predicted_price)^2
  )
)

#R-squared
r2 <- cor(
  test_data$Price_Lakh,
  predicted_price
)^2

print(paste("RMSE =", round(rmse, 2)))
print(paste("R² =", round(r2, 2)))

results <- data.frame(
  Actual = test_data$Price_Lakh,
  Predicted = round(predicted_price, 2)
)

head(results, 10)

#Price of a new car
new_car <- data.frame(
  Engine_CC = 1500,
  Mileage_kmpl = 20,
  Car_Age = 3,
  Brand = factor(
    "Hyundai",
    levels = levels(car_data$Brand)
  ),
  Fuel_Type = factor(
    "Petrol",
    levels = levels(car_data$Fuel_Type)
  )
)

new_price <- predict(
  linear_model,
  newdata = new_car
)

print(paste(
  "Predicted Price (Lakhs) =",
  round(new_price, 2)
))
```


**Inference**

The Linear Regression model predicts the price of a car as a continuous numerical value.
RMSE represents the difference between actual and predicted prices, while R² indicates how well the model explains the variation in car prices.
The trained model can also be used to estimate the price of a new car based on its engine, mileage, age, brand and fuel type.

```r
#LOGISTIC REGRESSION
car_data <- read.csv("/content/car_regression_classification_dataset_160.csv")

head(car_data)
str(car_data)
summary(car_data)

# Preprocessing
car_data$Brand <- as.factor(car_data$Brand)
car_data$Fuel_Type <- as.factor(car_data$Fuel_Type)
car_data$Car_Category <- as.factor(car_data$Car_Category)

# missing values
colSums(is.na(car_data))

#data into training and testing sets
set.seed(123)

train_index <- sample(
  1:nrow(car_data),
  size = 0.8 * nrow(car_data)
)

train_data <- car_data[train_index, ]
test_data <- car_data[-train_index, ]

#Logistic Regression model
logistic_model <- glm(
  Car_Category ~ Engine_CC + Mileage_kmpl +
    Car_Age + Brand + Fuel_Type,
  data = train_data,
  family = binomial
)

#model SUMMARY
summary(logistic_model)

#probabilities
probability <- predict(
  logistic_model,
  newdata = test_data,
  type = "response"
)

#probabilities into classes
predicted_class <- ifelse(
  probability >= 0.5,
  "Premium",
  "Regular"
)

predicted_class <- factor(
  predicted_class,
  levels = levels(car_data$Car_Category)
)

# Confusion Matrix
confusion_matrix <- table(
  Actual = test_data$Car_Category,
  Predicted = predicted_class
)

print(confusion_matrix)

# Accuracy
accuracy <- mean(
  predicted_class == test_data$Car_Category
)

#Precision
TP <- confusion_matrix["Premium", "Premium"]
FP <- confusion_matrix["Regular", "Premium"]

precision <- TP / (TP + FP)

#Recall
FN <- confusion_matrix["Premium", "Regular"]

recall <- TP / (TP + FN)

print(paste(
  "Accuracy =", round(accuracy, 2)
))

print(paste(
  "Precision =", round(precision, 2)
))

print(paste(
  "Recall =", round(recall, 2)
))

#actual and predicted classes
results <- data.frame(
  Actual = test_data$Car_Category,
  Predicted = predicted_class
)

head(results, 10)

#category of a new car
new_car <- data.frame(
  Engine_CC = 1500,
  Mileage_kmpl = 20,
  Car_Age = 3,
  Brand = factor(
    "Hyundai",
    levels = levels(car_data$Brand)
  ),
  Fuel_Type = factor(
    "Petrol",
    levels = levels(car_data$Fuel_Type)
  )
)

new_probability <- predict(
  logistic_model,
  newdata = new_car,
  type = "response"
)

new_category <- ifelse(
  new_probability >= 0.5,
  "Premium",
  "Regular"
)

print(paste(
  "Predicted Category =",
  new_category
))
```

**Inference**

The Logistic Regression model classifies cars into Premium and Regular categories using their characteristics.
Accuracy, precision, recall and the confusion matrix are used to evaluate the classification performance on the test data.
The model can also classify a new car by calculating its probability and assigning it to the appropriate category.
