Data Manipulation and Analysis Using R & Infering

```r
students <- read.csv("/content/student_99.csv")
```

```r
library(dplyr)

students %>%
  select(Name, Department, Marks)

students %>%
  filter(Marks > 75)
```

**Inference:**
-select() function displays only the required student details: Name, Department and Marks.
-filter() function identifies students who scored more than 75 marks.
it becomes easier to focus on selected columns

```r
employees <- read.csv("/content/employee_99.csv")
```

```r
library(dplyr)


employees %>%
  arrange(desc(Salary)) %>%
  head(5)
```

**Inference**

The employees are arranged from the highest salary to the lowest salary.
The first five records represent the employees with the highest salaries.
This makes it easy to identify the top-paid employees in the organization.

```r
library(dplyr)

students <- read.csv("/content/student_99.csv")

students %>%
  mutate(Result = ifelse(Marks >= 40, "Pass", "Fail"))
```

**Inference**

The mutate() function creates a new column called Result.
Students scoring 40 or above are classified as Pass, while those below 40 are classified as Fail.


```r
library(dplyr)

students <- read.csv("/content/student_99.csv")

students %>%
  group_by(Department) %>%
  summarise(
    Average_Marks = mean(Marks),
    Maximum_Marks = max(Marks),
    Minimum_Marks = min(Marks)
  )
```

```r
library(dplyr)

students <- read.csv("/content/student_99.csv")

students %>%
  group_by(Department) %>%
  summarise(
    Average_Marks = mean(Marks),
    Maximum_Marks = max(Marks),
    Minimum_Marks = min(Marks)
  )
```

**Inference**

AI&DS has the highest average marks among the departments.
CSE follows with a relatively high average
EEE has the lowest average.

```r
library(data.table)
employees<- as.data.table(read.csv("/content/dept_Salary.csv"))

employees[, .(Average_Salary = mean(Salary)), by = Department]
```

```r
avg_salary <- employees[, .(Average_Salary = mean(Salary)),
                        by = Department]

avg_salary[which.max(Average_Salary)]
```

**Inference**

The average salary is calculated separately for each department using data.table.
The department with the highest average salary is identified using which.max().

```r
library(tidyr)

students <- read.csv("/content/marks_subjects.csv")

students_long <- students %>%
  pivot_longer(
    cols = Mathematics:Computer_Science,
    names_to = "Subject",
    values_to = "Marks"
  )

students_long
```

**Inference**

The original dataset stores each subject as a separate column, which is called wide format.
pivot_longer() converts these subject columns into a single Subject column and their values into a Marks column.
