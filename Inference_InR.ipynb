# Create a student dataset with Name, Marks, and Attendance. Using dplyr, create a new column called Result that contains "Pass" if marks ≥ 40 and "Fail" otherwise. Provide an inference about the results.

# Create a dataset containing students and their marks in Mathematics, Science, English, and Computer Science in wide format. Use pivot_longer() to convert it into long format. Display the resulting dataset and provide an inference about the change in data structure.

```r
library(tidyr)

students <- read.csv("/content/marks_subjects.csv")

students_long <- students %>%
  pivot_longer(
    cols = Mathematics:Computer_Science,
    names_to = "Subject",
    values_to = "Marks"
  )

students_long
```

**Inference**

The original dataset stores each subject as a separate column, which is called wide format.
pivot_longer() converts these subject columns into a single Subject column and their values into a Marks column.

# Create a long-format dataset containing Student, Subject, and Marks. Use pivot_wider() to convert it into wide format where each subject becomes a separate column. Explain the difference between the two formats.

```r
library(tidyr)

students <- read.csv("/content/LONG_Std.csv")

students_wide <- students %>%
  pivot_wider(
    names_from = Subject,
    values_from = Marks
  )

students_wide
```

**Inference**

pivot_wider() changes the dataset from long format to wide format.
The subjects that were stored as values in the Subject column become separate columns.
Wide format makes it easier to compare

# Create a dataset containing monthly sales of three products. Use reshape2 to convert the data from wide format to long format using melt(), and then convert it back using dcast(). Explain the purpose of reshaping.

```r
install.packages("reshape2")
library(reshape2)
```

```r
library(reshape2)

sales <- read.csv("/content/monthlysales99.csv")

sales
```

```r
sales_long <- melt(
  sales,
  id.vars = "Month",
  variable.name = "Product",
  value.name = "Sales"
)

sales_long
```

```r
sales_wide <- dcast(
  sales_long,
  Month ~ Product,
  value.var = "Sales"
)

sales_wide
```

**Inference**

melt() converts the sales data from wide format into long format by combining the product columns.
dcast() converts the long data back into its original wide structure.
